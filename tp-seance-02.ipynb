**Chronométrer** : `time.perf_counter()` donne l'heure en secondes ; la **différence** entre deux mesures est la **durée** d'un calcul. On compare ici une addition de deux millions de valeurs : avec une boucle sur des listes, puis avec des tableaux.

```python
import time

n = 1_000_000
liste_a, liste_b = list(range(n)), list(range(n))
tab_a, tab_b = np.arange(n), np.arange(n)

debut = time.perf_counter()
somme_liste = [liste_a[i] + liste_b[i] for i in range(n)]
t_liste = time.perf_counter() - debut

debut = time.perf_counter()
somme_tab = tab_a + tab_b
t_tab = time.perf_counter() - debut

print("liste   :", round(t_liste * 1000, 1), "ms")
print("tableau :", round(t_tab * 1000, 2), "ms")
print("le tableau est", round(t_liste / t_tab), "fois plus rapide")
```

# Série 2 · Lire et découper ⭐ → ⭐⭐
_Diapos 20 à 25 et 28._

**Cours en bref**
- **Indice** : `a[i]` (en 2-D : `a[ligne, colonne]`) ; un indice **négatif** compte depuis la fin (`-1` : la dernière case).
- **Tranche** : `a[début:fin:pas]` inclut le début et **exclut** la fin ; en 2-D, une tranche **par axe**, séparées par une virgule : `a[lignes, colonnes]`.
- **Masque** : une condition (`a > 5`) donne un tableau de `True` / `False` ; `a[a > 5]` garde les cases vraies. On combine avec `&` (et), `|` (ou), `~` (non), **entre parenthèses**.
- **Axe** : `a.mean(axis=0)` donne une valeur **par colonne**, `axis=1` une valeur **par ligne**.

```python
print(notes[0])              # la ligne d'Ali
print(notes[0, 1])           # Ali en Stats
print(notes[-1, -1])         # la dernière note
print(notes[:, 0])           # toute la colonne Python
print(notes[1:4, 1:3])       # un bloc : lignes 1 à 3, colonnes 1 et 2
print(notes[notes >= 17])    # les notes à partir de 17
print(notes.mean(axis=0))    # une moyenne par matière
print(notes.mean(axis=1))    # une moyenne par étudiant
```

# Série 4 · Calculer sans boucle ⭐⭐
_Diapos 30 à 35._

**Cours en bref**
- Les opérations (`+`, `-`, `*`, `/`, `**`, `%`) s'appliquent **case par case**, sans boucle : c'est la **vectorisation**.
- Les **fonctions universelles** (`np.sqrt`, `np.abs`, `np.exp`, `np.around`, `np.floor`, `np.ceil`, `np.log`, `np.log10`, `np.log2`...) s'appliquent à **chaque case**. `np.around` arrondit au **pair** le plus proche quand on tombe pile entre deux entiers.
- `np.cumsum` donne les **sommes cumulées**, `np.diff` les **écarts** entre cases voisines.
- **Broadcasting** : deux tableaux de formes différentes se combinent si, de droite à gauche, leurs dimensions sont égales ou valent 1 : un `(8, 4)` moins un `(4,)` marche.
- Une case **manquante** s'écrit `np.nan` ; elle **contamine** `mean`, alors que `np.nanmean` et `np.nanmedian` l'ignorent. `np.isnan(a)` repère les `nan`.

```python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
print(a + b, a * b, b / a, a ** 2, b % 3)

print(np.sqrt(a))
print(np.around(np.array([2.567, 3.5, 4.5]), 1), np.around(np.array([2.5, 3.5, 4.5])))
print(np.log(np.array([1.0, np.e])), np.log10(np.array([1, 10, 100])))
print(np.cumsum(b), np.diff(b))

print(notes - notes.mean(axis=0))        # (8, 4) - (4,) : le vecteur est étiré sur les 8 lignes

t = np.array([20.0, np.nan, 22.0])
print(t.mean(), np.nanmean(t), np.isnan(t))
```

# Série 7 · Pour aller plus loin ⭐⭐ → ⭐⭐⭐
_Support de cours, chapitres 1, 4 et 5 : ces notions ne sont pas dans les diapos._

**Cours en bref**
- **Assembler** : `np.vstack((a, b))` empile en **hauteur**, `np.hstack((a, b))` colle **côte à côte**, `np.concatenate((a, b))` assemble selon un axe.
- **Découper** : `np.split(a, 3)` en 3 morceaux **égaux** (sinon, erreur) ; `np.array_split(a, 3)` accepte des morceaux **inégaux**.
- **Ensembles** : `np.unique(a)` (valeurs distinctes, triées), `np.union1d`, `np.intersect1d`, `np.setdiff1d`.
- **Chercher dans un tableau trié** : `np.searchsorted(seuils, valeurs, side="right")` donne le **rang** de chaque valeur parmi les seuils.
- **Algèbre linéaire** : `A.T` (transposée), `np.dot(A, B)` (produit matriciel), `np.linalg.det(A)` (déterminant), `np.linalg.solve(A, b)` (résout `A x = b`).

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
print(np.vstack((a, b)))                    # 3 lignes, 2 colonnes
print(np.hstack((a, a)))                    # 2 lignes, 4 colonnes
print(np.concatenate((np.arange(3), np.arange(10, 13))))

print(np.split(np.arange(6), 3))            # trois morceaux de 2
print(np.array_split(np.arange(7), 3))      # des morceaux de 3, 2 et 2

print(np.unique(np.array([3, 1, 3, 2, 1])))
print(np.union1d([1, 2, 3], [3, 4]), np.intersect1d([1, 2, 3], [3, 4]), np.setdiff1d([1, 2, 3], [3, 4]))

seuils = np.array([10, 12, 14])
print(np.searchsorted(seuils, [9, 10, 13, 20], side="right"))    # le rang de chaque valeur parmi les seuils

A = np.array([[2, 1], [1, 3]])
print(A.T)                                  # la transposée
print(np.dot(A, A))                         # le produit matriciel
print(np.linalg.det(A))                     # le déterminant
print(np.linalg.solve(A, np.array([8, 13])))    # la solution de A x = b
```

# Série 8 · Lire et écrire des fichiers ⭐⭐ → ⭐⭐⭐
_Fiche résumé NumPy : « Saving & Loading »._

**Cours en bref**
- On range les fichiers d'essai dans un dossier : `dossier = Path("fichiers_tp")` puis `dossier.mkdir(exist_ok=True)` (le crée s'il n'existe pas). Pour **voir** le contenu d'un fichier texte : `chemin.read_text()`.
- **Écrire un fichier texte** : `np.savetxt(chemin, tableau, delimiter=",", fmt="%d", header="a,b", comments="")` : séparateur, **format** des nombres (`%d` entier, `%g` décimal court, `%.2f` deux décimales), **en-tête** (sans `comments=""`, NumPy ajoute un `#` devant).
- **Lire un fichier texte** : `np.loadtxt(chemin, delimiter=",", skiprows=1)` (`skiprows=1` saute l'en-tête). Une **case vide** la fait échouer ; `np.genfromtxt(chemin, delimiter=",", skip_header=1)` la remplace par `nan`.
- **Binaire** : `np.save(chemin, tableau)` écrit un `.npy` (un tableau, sans rien perdre : type et forme compris) ; `np.load(chemin)` le relit ; `np.savez(chemin, nom1=tableau1, nom2=tableau2)` range **plusieurs** tableaux dans un `.npz`.

```python
from pathlib import Path

dossier = Path("fichiers_tp")
dossier.mkdir(exist_ok=True)                # crée le dossier s'il n'existe pas

# 1. un tableau dans un fichier texte, puis on regarde le fichier tel quel
petit = np.array([[1, 2.5], [3, 4.0]])
np.savetxt(dossier / "petit.csv", petit, delimiter=",", fmt="%g", header="a,b", comments="")
print((dossier / "petit.csv").read_text())

# 2. on le relit
relu = np.loadtxt(dossier / "petit.csv", delimiter=",", skiprows=1)
print(relu, relu.shape, relu.dtype)

# 3. un fichier avec une case vide : loadtxt refuse, genfromtxt met nan
(dossier / "trou.csv").write_text("x,y\n1,2\n3,\n5,6\n", encoding="utf-8")
try:
    np.loadtxt(dossier / "trou.csv", delimiter=",", skiprows=1)
except ValueError as e:
    print("loadtxt refuse la case vide :", e)
print(np.genfromtxt(dossier / "trou.csv", delimiter=",", skip_header=1))

# 4. en binaire : un tableau (.npy), puis plusieurs (.npz)
np.save(dossier / "petit.npy", petit)
print(np.load(dossier / "petit.npy"))
np.savez(dossier / "deux.npz", a=petit, b=np.arange(3))
with np.load(dossier / "deux.npz") as archive:
    print(archive["a"], archive["b"])
```
